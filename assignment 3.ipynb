### Tossing a Coin 10,000 Times

```python
import random

trials = 10000
heads = 0
tails = 0

for _ in range(trials):
    if random.choice(['H', 'T']) == 'H':
        heads += 1
    else:
        tails += 1

print("Experimental Probability of Heads:", heads / trials)
print("Experimental Probability of Tails:", tails / trials)
```

### Rolling Two Dice – Probability of Sum = 7

```python
count_sum_7 = 0
trials = 10000

for _ in range(trials):
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)
    if die1 + die2 == 7:
        count_sum_7 += 1

print("Probability of getting sum 7:", count_sum_7 / trials)
```

### Probability of Getting At Least One “6” in 10 Rolls

```python
def probability_at_least_one_six(trials=10000):
    success = 0

    for _ in range(trials):
        rolls = [random.randint(1, 6) for _ in range(10)]
        if 6 in rolls:
            success += 1

    return success / trials

print("Estimated Probability:", probability_at_least_one_six())
```

### Conditional Probability & Bayes’ Theorem

```python
colors = ['Red'] * 5 + ['Green'] * 7 + ['Blue'] * 8

prev_blue = 0
blue_then_red = 0

trials = 1000
previous = None

for _ in range(trials):
    current = random.choice(colors)

    if previous == 'Blue':
        prev_blue += 1
        if current == 'Red':
            blue_then_red += 1

    previous = current

conditional_prob = blue_then_red / prev_blue
print("P(Red | Previous Blue):", conditional_prob)
```

### Discrete Random Variable Simulation

```python
import numpy as np

values = [1, 2, 3]
probabilities = [0.25, 0.35, 0.4]

sample = np.random.choice(values, size=1000, p=probabilities)

print("Mean:", np.mean(sample))
print("Variance:", np.var(sample))
print("Standard Deviation:", np.std(sample))
```

### Continuous Random Variable – Exponential Distribution

```python
import matplotlib.pyplot as plt

samples = np.random.exponential(scale=5, size=2000)

plt.hist(samples, bins=40, density=True, alpha=0.6)

x = np.linspace(0, max(samples), 1000)
pdf = (1/5) * np.exp(-x/5)
plt.plot(x, pdf)

plt.title("Exponential Distribution (Mean = 5)")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

### Central Limit Theorem Simulation

```python
uniform_data = np.random.uniform(0, 1, 10000)

sample_means = []
for _ in range(1000):
    sample = np.random.choice(uniform_data, size=30)
    sample_means.append(np.mean(sample))

plt.hist(uniform_data, bins=40, density=True, alpha=0.5)
plt.title("Uniform Distribution")
plt.show()

plt.hist(sample_means, bins=40, density=True, alpha=0.7)
plt.title("Distribution of Sample Means (CLT)")
plt.show()
```
